```python
import numpy as np
```

```python
#Importe de siniestros anuales por poliza.
siniestros = np.array([450,800,1200,1500,2300,5000])
exposiciones = np.array([1,1,1,0.5,2,1])
pesos = np.array([0.1,0.15,0.2,0.1,0.25,0.2])
```

Funciones estadisticas de numpy:

np.mean()
np.median()
np.sum()
np.std()
np.var()
np.percentil()
np.quantile()

```python
print(np.mean(siniestros))
print(np.median(siniestros))
print(np.sum(siniestros))
print(np.std(siniestros))
print(np.var(siniestros))
print(np.percentile(siniestros,95))
print(np.quantile(siniestros,0.5))
```

Segmentacion de riesgos: Clasificacion automatica de mi riesgo en funcion de mi prima:
 np.where()

```python
riesgo = np.where(siniestros>2000,"Alto","Normal")
print(riesgo)
```

np.clip() -> Creamos una prima con limite de cobertura

```python
pago_limitado =np.clip(siniestros,0,3000)
print(pago_limitado)
```

np.maximun() -> deducir x euros de la poliza

```python
pago_deducido=np.maximum(siniestros -500,0)
print(pago_deducido)
```

Media ponderada np.dot(data, ponderacion)

```python
print(np.dot(siniestros,pesos))
```

pagos acumulados - Evolucion acumulado del coste
    np.cumsum(vector)

```python
print(np.cumsum(siniestros))
```

Frecuencia de siniestros- funcion de poisson

```python
print(np.random.poisson(0.8,10)) # poisson(lambda,tamaño de la muestra)
```

np.power() - Ej: calcular valor presente a dos años con un 3% anual

```python
VA=siniestros/np.power(1.03,2)
print(VA)
```

Evolucion de la siniestralidad en el tiempo usando como referencia la cantidad de siniestros
np.polyfit(años,data)

```python
años = np.arange(len(siniestros))
tendencia = np.polyfit(años,siniestros,0)
print(tendencia)
```

Ranking de riesgos - Identificar las polizas más peligrosas - np.argsort()

```python
rank=np.argsort(siniestros)[::-1]
print(rank)
```
